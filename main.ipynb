```python
# ============================================================
# Create utils.py
# ============================================================

utils_code = r'''
"""
============================================================
ARIA Research Prototype
Utility Functions
============================================================
"""

import hashlib
import json
import secrets
import time
import uuid


class CryptoUtils:

    @staticmethod
    def sha256(data):

        if isinstance(data, dict):
            data = json.dumps(data, sort_keys=True)

        if not isinstance(data, str):
            data = str(data)

        return hashlib.sha256(data.encode()).hexdigest()


    @staticmethod
    def random_token():

        return str(uuid.uuid4())


    @staticmethod
    def nonce(length=16):

        return secrets.token_hex(length)


class TimeUtils:

    @staticmethod
    def now():

        return time.time()


    @staticmethod
    def timestamp():

        return int(time.time())


class JSONUtils:

    @staticmethod
    def save(data, filename):

        with open(filename, "w") as fp:

            json.dump(data, fp, indent=4)


    @staticmethod
    def load(filename):

        with open(filename, "r") as fp:

            return json.load(fp)

'''

with open("src/utils.py","w") as fp:

    fp.write(utils_code)

print("✓ utils.py successfully created.")
```

```python
# ============================================================
# Create models.py (Part 1 - Enumerations)
# ============================================================

models_code = r'''
"""
============================================================
ARIA Research Prototype
Common Data Models
============================================================
"""

from enum import Enum


# ============================================================
# Fault Categories
# ============================================================

class FaultCategory(Enum):

    RESOURCE = "Resource"

    APPLICATION = "Application"

    NETWORK = "Network"

    SECURITY = "Security"

    SYSTEM = "System"


# ============================================================
# Fault Severity
# ============================================================

class FaultSeverity(Enum):

    LOW = 1

    MEDIUM = 2

    HIGH = 3

    CRITICAL = 4


# ============================================================
# Remediation Actions
# ============================================================

class RemediationAction(Enum):

    NONE = "none"

    CLEAR_CACHE = "clear_cache"

    RESTART_APPLICATION = "restart_application"

    RESTART_SERVICE = "restart_service"

    CLEANUP_STORAGE = "cleanup_storage"

    RESET_NETWORK = "reset_network"

    APPLY_PATCH = "apply_patch"

    CLOUD_ASSISTED = "cloud_assisted"

    HUMAN_APPROVAL = "human_approval"


# ============================================================
# Authorization Decision
# ============================================================

class AuthorizationDecision(Enum):

    ALLOW = "ALLOW"

    ESCALATE = "ESCALATE"

    HUMAN_APPROVAL = "HUMAN_APPROVAL"

    REJECT = "REJECT"

'''

with open("src/models.py", "w") as fp:

    fp.write(models_code)

print("✓ models.py (Part 1) created successfully.")
```

```python
# ============================================================
# Append Diagnostic Engine
# ============================================================

diagnostic_code = r'''

from utils import TimeUtils


# ============================================================
# Diagnostic Engine
# ============================================================

class DiagnosticEngine:
    """
    Identifies the dominant fault affecting the device based
    on runtime telemetry.
    """

    @staticmethod
    def diagnose(state):

        # ----------------------------------------------------
        # Resource Faults
        # ----------------------------------------------------

        if state.memory_utilization >= 85:

            return FaultEvent(

    category=FaultCategory.RESOURCE,

    fault_code="MEMORY_HIGH",

    severity=FaultSeverity.HIGH,

    description="High memory utilization",

    risk_score=0.85,

    timestamp=TimeUtils.now()

)

        if state.storage_utilization >= 90:

            return FaultEvent(

                category=FaultCategory.RESOURCE,

                severity=FaultSeverity.HIGH,

                description="Storage almost full",

                risk_score=0.90,

                timestamp=TimeUtils.now()

            )

        # ----------------------------------------------------
        # Application Faults
        # ----------------------------------------------------

        if not state.application_running:

            return FaultEvent(

    category=FaultCategory.RESOURCE,

    fault_code="STORAGE_HIGH",

    severity=FaultSeverity.HIGH,

    description="Storage almost full",

    risk_score=0.90,

    timestamp=TimeUtils.now()

)

        # ----------------------------------------------------
        # Network Faults
        # ----------------------------------------------------

        if state.network_latency >= 250:

            return FaultEvent(

    category=FaultCategory.NETWORK,

    fault_code="NETWORK_LATENCY",

    severity=FaultSeverity.MEDIUM,

    description="High network latency",

    risk_score=0.65,

    timestamp=TimeUtils.now()

)

        # ----------------------------------------------------
        # Security Faults
        # ----------------------------------------------------

        if state.security_alert:

            return FaultEvent(

    category=FaultCategory.SECURITY,

    fault_code="SECURITY_ALERT",

    severity=FaultSeverity.CRITICAL,

    description="Security alert detected",

    risk_score=0.99,

    timestamp=TimeUtils.now()

)

        # ----------------------------------------------------
        # Healthy Device
        # ----------------------------------------------------

        return None

'''

with open("src/device_agent.py", "a") as fp:

    fp.write(diagnostic_code)

print("✓ DiagnosticEngine appended.")
```

```python

# ============================================================
# Action Selector
# ============================================================

class ActionSelector:
    """
    Maps diagnosed faults to remediation actions.
    """

    @staticmethod
    def select(fault):

        if fault is None:

            return RemediationAction.NONE

        mapping = {

            "MEMORY_HIGH": RemediationAction.CLEAR_CACHE,

            "STORAGE_HIGH": RemediationAction.CLEANUP_STORAGE,

            "APPLICATION_CRASH": RemediationAction.RESTART_APPLICATION,

            "NETWORK_LATENCY": RemediationAction.RESET_NETWORK,

            "SECURITY_ALERT": RemediationAction.CLOUD_ASSISTED,

        }

        return mapping.get(

            fault.fault_code,

            RemediationAction.NONE

        )
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# IEEE Figure 10: 95% Confidence Interval
# ==========================================================

# Load confidence interval file
df = pd.read_csv("confidence_interval.csv")

# Extract values
scenarios = df["Scenario"]

mean = df["Mean"]

lower = np.maximum(df["Lower95"], 0)     # clip negative latency
upper = df["Upper95"]

# Error bar lengths
lower_err = mean - lower
upper_err = upper - mean

fig, ax = plt.subplots(figsize=(7.2,4.2))

ax.errorbar(
    scenarios,
    mean,
    yerr=[lower_err, upper_err],
    fmt='o',
    markersize=7,
    linewidth=2,
    elinewidth=1.8,
    capsize=5,
    capthick=1.8
)

# Axis labels
ax.set_ylabel("Mean Recovery Latency (ms)", fontsize=11)
ax.set_xlabel("Benchmark Scenario", fontsize=11)

# Grid
ax.grid(True, linestyle='--', alpha=0.35)

# Cleaner IEEE style
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Better x labels
plt.xticks(rotation=15)

plt.tight_layout()

plt.savefig(
    "confidence_interval_ieee.pdf",
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    "confidence_interval_ieee.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# Figure: Latency Component Breakdown
# ============================================================

scenarios = [
    "Memory",
    "Storage",
    "Application",
    "Network",
    "Security"
]

diagnosis = np.array([
    0.007,
    0.001,
    0.001,
    0.032,
    0.059
])

authentication = np.array([
    0.000,
    0.000,
    0.000,
    0.030,
    0.025
])

cloud = np.array([
    0.000,
    0.000,
    0.000,
    0.017,
    0.010
])

x = np.arange(len(scenarios))
width = 0.60

fig, ax = plt.subplots(figsize=(7.4,4.6))

ax.bar(
    x,
    diagnosis,
    width,
    label="Diagnosis"
)

ax.bar(
    x,
    authentication,
    width,
    bottom=diagnosis,
    label="Authentication"
)

ax.bar(
    x,
    cloud,
    width,
    bottom=diagnosis+authentication,
    label="Cloud Recommendation"
)

ax.set_xticks(x)
ax.set_xticklabels(scenarios)

ax.set_ylabel("Latency (ms)")
ax.set_xlabel("Benchmark Scenario")

ax.grid(axis='y', linestyle='--', alpha=0.35)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.legend()

plt.tight_layout()

plt.savefig(
    "latency_breakdown_components.pdf",
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    "latency_breakdown_components.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================================
# IEEE Figure: Recovery Workflow Latency Contribution
# ==========================================================

components = [
    "Diagnosis",
    "Authentication",
    "Cloud\nRecommendation"
]

latency = np.array([
    0.005727,
    0.005607,
    0.002780
])

# Percentage contribution
percent = latency / latency.sum() * 100

fig, ax = plt.subplots(figsize=(6.6,4.2))

bars = ax.barh(
    components,
    latency,
    height=0.55
)

ax.set_xlabel("Average Latency (ms)", fontsize=11)

ax.grid(axis='x', linestyle='--', alpha=0.35)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

for b, p in zip(bars, percent):
    ax.text(
        b.get_width()+0.00008,
        b.get_y()+b.get_height()/2,
        f"{p:.1f}%",
        va="center",
        fontsize=10
    )

plt.tight_layout()

plt.savefig(
    "workflow_latency_contribution.pdf",
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    "workflow_latency_contribution.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================================
# Figure 7
# Benchmark Coverage Analysis
# IEEE IoT Version
# ==========================================================

categories = [
    "System Faults",
    "Security Attacks",
    "Application\nServices",
    "Network\nInfrastructure"
]

autonomous = np.array([5, 1, 3, 2])
cloud = np.array([0, 6, 2, 1])

y = np.arange(len(categories))
h = 0.34

fig, ax = plt.subplots(figsize=(7.4, 4.3))

# -------------------------------
# Bars
# -------------------------------
bars1 = ax.barh(
    y - h/2,
    autonomous,
    height=h,
    label="Autonomous"
)

bars2 = ax.barh(
    y + h/2,
    cloud,
    height=h,
    label="Cloud-Assisted"
)

# -------------------------------
# Value labels
# -------------------------------
for b in bars1:
    w = b.get_width()
    if w > 0:
        ax.text(
            w + 0.08,
            b.get_y() + b.get_height()/2,
            f"{int(w)}",
            va="center",
            fontsize=10
        )

for b in bars2:
    w = b.get_width()
    if w > 0:
        ax.text(
            w + 0.08,
            b.get_y() + b.get_height()/2,
            f"{int(w)}",
            va="center",
            fontsize=10
        )

# -------------------------------
# Axes
# -------------------------------
ax.set_xlabel(
    "Number of Scenarios",
    fontsize=11
)

ax.set_yticks(y)
ax.set_yticklabels(
    categories,
    fontsize=10
)

ax.set_xlim(0,7)

ax.set_xticks(np.arange(0,8,1))

# -------------------------------
# Grid
# -------------------------------
ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.35
)

# -------------------------------
# IEEE style
# -------------------------------
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    loc="upper right",
    frameon=False,
    fontsize=10
)

plt.tight_layout(pad=0.4)

plt.savefig(
    "benchmark_coverage.pdf",
    dpi=600,
    bbox_inches="tight"
)

plt.savefig(
    "benchmark_coverage.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```
